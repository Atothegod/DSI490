```python
import os
import dspy
from pydantic import BaseModel, Field
from typing import Literal
from dotenv import load_dotenv
load_dotenv() 
```

```python
api_key = os.getenv("GEMINI_API_KEY") or os.getenv("GOOGLE_API_KEY") or os.getenv("API_KEY")
model = os.getenv("GEMINI_MODEL", "gemini/gemini-2.5-flash")
lm = dspy.LM(model=model, api_key=api_key)
dspy.configure(lm=lm)
```

```python
class HistoryMessage(BaseModel):
    role: Literal["system", "user", "assistant"]
    content: str = Field(min_length=1)


class ChatRequest(BaseModel):
    input_text: str = Field(min_length=1)
    history: list[HistoryMessage] = Field(default_factory=list)


class ChatResponse(BaseModel):
    output_text: str
    model: str
```

```python
class AgentSignature(dspy.Signature):
    """You are Dr.Wasit a helpful assistant that answers questions about the course DSI490. You have access to the session history and the latest user message. Use this information to provide a helpful answer to the user."""
    input_text: str = dspy.InputField(desc="The latest user message.")
    output_text: str = dspy.OutputField(desc="Helpful answer to the latest user message.")
```

```python
agent = dspy.Predict(AgentSignature)
```

```python
output = agent(input_text="สวัสดีครับ")
```
